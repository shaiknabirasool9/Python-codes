```python

#Task 5 — Price & Quantity Investigation
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("ECommerce_Orders_.csv")

# Create Price_Category
conditions = [
    df["Unit_Price"] < 2000,
    (df["Unit_Price"] >= 2000) & (df["Unit_Price"] <= 10000),
    (df["Unit_Price"] > 10000) & (df["Unit_Price"] <= 30000),
    df["Unit_Price"] > 30000
]

choices = [
    "Budget",
    "Mid Range",
    "Premium",
    "Luxury"
]

df["Price_Category"] = np.select(
    conditions,
    choices,
    default="Unknown"
)

# Find Luxury products with Quantity > 1
luxury_products = df[
    (df["Price_Category"] == "Luxury") &
    (df["Quantity"] > 1)
]

# Sort by Unit_Price descending
result = luxury_products.sort_values(
    by="Unit_Price",
    ascending=False
)

# Display required columns
result = result[
    ["Product", "Category", "Quantity", "Unit_Price", "Price_Category"]
]

print(result)
```

```python
# Task 10 — Complete Analyst Challenge
# Dataset: E-Commerce Orders
# Concepts: Combined Day 1–8 workflow; no groupby() or aggregation
# •	Create Revenue = Quantity × Unit_Price.
# •	Create Discount_Rate: Revenue >= 50000 = 15%; Revenue >= 20000 = 10%; otherwise 5%.
# •	Calculate Discount_Amount and Final_Revenue.
# •	Create Customer_Status: Rating >= 4 = Satisfied; otherwise Needs Attention.
# •	Find orders where Category = Electronics AND Final_Revenue > 20000 AND Rating >= 4 AND Payment_Mode = UPI.
# •	Sort qualifying orders by Final_Revenue descending.
# •	Return Order_ID, Customer, City, Product, Quantity, Revenue, Discount_Amount, Final_Revenue, and Rating.
# •	Create Order_Label such as: ORD001 - Laptop - ₹55250
```

```python
df=pd.read_csv("ECommerce_Orders_.csv")
df
```

```python
df["Revenue"]=df["Quantity"]*df["Unit_Price"]
cond=[
    df["Revenue"]>=50000,
    (df["Revenue"]>=20000)&(df["Revenue"]<50000)
]
choice=[0.15,0.10]
df["Discount_Amount"]=np.select(cond,choice,default=[0.5])
df["Final_Revenue"] = df["Revenue"] - df["Discount_Amount"]
df["Customer_Status"]=np.where(df["Rating"] >= 4,"Satisfied","Needs Attention")
df
```

```python
qualifying_filter = (
    (df["Category"] == "Electronics")
    & (df["Final_Revenue"] > 20000)
    & (df["Rating"] >= 4)
    & (df["Payment_Mode"] == "UPI")
)
df[qualifying_filter]
```

```python
df=df.sort_values(by="Final_Revenue",ascending=False)
df
```

```python
df["Order_Label"] = (
    df["Order_ID"].astype(str)
    + " - "
    + df["Product"].astype(str)
    + " - ₹"
    + df["Final_Revenue"].round(2).astype(str)
)
df
```
