```python

Design a base class Employee that encapsulates common attributes (name, age, employee ID) and methods (e.g., display_details).
Create derived classes Professor and Lecturer that inherit from Employee and add their specific attributes and methods.
Implement encapsulation by ensuring that attributes are accessed and modified through methods.
Demonstrate inheritance by creating instances of Professor and Lecturer and calling both inherited and specific methods.
Write a main function to test your classes with at least two instances of each type, showcasing their unique and shared functionalities.
Ensure your code is well-documented with comments and follows professional coding standards.
```

```python
class Employee:
  def __init__(self,name,age,employee_id):
    self.name = name
    self.age = age
    self.employee_id = employee_id
  def display_details(self):
    print(f"Name: {self.name}")
    print(f"Age: {self.age}")
    print(f"Employee ID: {self.employee_id}")

class Professor(Employee):
  def __init__(self,name,age,employee_id,department):
    super().__init__(name,age,employee_id)
    self.department = department
  def display_details(self):
    super().display_details()
    print(f"Department: {self.department}")

class Lecturer(Employee):
  def __init__(self,name,age,employee_id,department):
    super().__init__(name,age,employee_id)
    self.department = department
    def display_details(self):
      super().display_details()
      print(f"Department: {self.department}")

print("*"*15)
le = Lecturer("Raju",22,2001,"Teaching")
le.display_details()
print("*"*15)
pr = Professor("Raja",23,2002,"Trainer")
pr.display_details()

```
